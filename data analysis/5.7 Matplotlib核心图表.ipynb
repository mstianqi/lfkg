```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']

figure, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 5), dpi=100)

# 制作柱状图bar
countries = ['国1', '国2', '国3', '国4']
values = [10, 20, 30, 40]
axes1:Axes = axes[0]
axes1.bar(countries, values, color='red',width=0.5)
axes1.set_title("柱状图对比", fontsize=20)
axes1.set_xlabel('国家')
axes1.set_ylabel('数量')

# 制作饼图pie
countries2 = ['国1', '国2', '国3', '国4']
values2 = [20, 20, 50, 40]
axes2:Axes = axes[1]
axes2.pie(values2, labels=countries2, autopct='%1.1f%%')
axes2.set_title("饼图对比", fontsize=20)
axes2.legend(loc='lower left', bbox_to_anchor=(0, -0.1))
```
