```python
import pandas as pd

# 创建数据
data = {
    '产品类别': ['电子产品', '电子产品', '家用电器', '家用电器', '服装鞋帽',
              '服装鞋帽', '食品饮料', '食品饮料', '电子产品', '家用电器'],
    '产品名称': ['智能手机', '蓝牙耳机', '电饭煲', '吸尘器', '运动鞋',
              '羽绒服', '牛奶', '矿泉水', '平板电脑', '微波炉'],
    '销售数量': [12, 25, 8, 5, 20, 10, 30, 50, 6, 7],
    '单价': [2999, 199, 499, 1299, 399, 899, 65, 3, 4999, 799]
}

df = pd.DataFrame(data)

# 写入CSV文件
df.to_csv('data/sales.csv',
          index=False,           # 不保存行索引
          encoding='utf-8-sig',  # 编码格式（含BOM，Excel打开不乱码）
          float_format='%.2f')   # 浮点数保留2位小数
```

```python
# 增加一行并写入
df2 = pd.read_csv('data/sales2.csv', usecols=['产品名称','销售数量','单价'])
df2['销售金额'] = df2['单价'] * df2['销售数量']
df2.to_csv('data/sales2.csv', index=False, encoding='utf-8-sig', float_format='%.2f')
```
