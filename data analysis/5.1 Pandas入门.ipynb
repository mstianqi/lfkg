```python
import pandas as pd

df = pd.DataFrame([
    {"姓名": "小明", "语文": 90, "数学": 80},
    {"姓名": "小红", "语文": 80, "数学": 85},
    {"姓名": "小强", "语文": 70, "数学": 60}
])

print(f"语文最高分：{df['语文'].max()}，最低分：{df['语文'].min()}，平均分：{df['语文'].mean():.2f}")
print(f"数学最高分：{df['数学'].max()}，最低分：{df['数学'].min()}，平均分：{df['数学'].mean():.2f}")

# DataFrame常用属性
df.index
df.columns
df.values
df.size
df.dtypes
df.shape
```
