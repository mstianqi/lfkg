```python
import pandas as pd

df = pd.read_csv('data/sales.csv')
df.sort_values('单价')
df.sort_values(['单价','销售数量'], ascending=[True, False])
```

```python
df2 = pd.read_csv('data/sales.csv')
df2['销售金额'] = df['单价'] * df['销售数量']
# 根据产品类别分组，统计销售数量
df2.groupby('产品类别')['销售数量'].sum()
# 根据产品类别分组，统计订单数量
df2.groupby('产品类别')['销售数量'].count()

df2.groupby('产品类别')['单价'].min()
df2.groupby('产品类别').agg({'销售数量':'sum', '单价':'mean'})
```
